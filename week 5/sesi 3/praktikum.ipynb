```python
import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import pdist, squareform
import matplotlib.pyplot as plt

# 1. Dataset sederhana
data = np.array([
    [1.0,1.0],  # a
    [1.5, 1.5], # b
    [5.0, 5.0], # c
    [3.0, 4.0], # d
    [4.0,4.0],      # e
    [3.0, 3.5]  # f

])

# 2. Hitung Distance Matrix (Euclidean Distance)
distance_matrix = pdist(data, metric='euclidean')
distance_df = pd.DataFrame(squareform(distance_matrix),
                            columns=['a', 'b', 'c', 'd', 'e', 'f'],
                            index=['a', 'b', 'c', 'd', 'e', 'f'])

# 3. Tampilkan Distance Matrix
print(" === Distance Matrix (Euclidean) === ")
print(distance_df)

# 4. Hierarchical Clustering (Single Linkage)
linked = linkage(data, method='single')

# 5. PLot Dendrogram
plt.figure(figsize=(12, 8))
plt.title("Hierarchical Clustering Dendrogram (Single Linkage)")
dendrogram(linked, labels=['a', 'b', 'c', 'd', 'e', 'f'], leaf_rotation=90)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.show()
```

penjelasan kode ini mengimpor beberapa library yang diperlukan untuk melakukan proses Hierarchical Clustering. numpy digunakan untuk membuat dataset dalam bentuk array, pandas digunakan untuk mengolah data dalam bentuk tabel, linkage dan dendrogram dari scipy.cluster.hierarchy digunakan untuk melakukan proses Hierarchical Clustering dan menampilkan dendrogram, sedangkan pdist dan squareform digunakan untuk menghitung jarak antar data. matplotlib.pyplot digunakan untuk menampilkan grafik. Dataset terdiri dari enam titik data yang diberi label a sampai f dengan dua atribut atau koordinat.
digunakan untuk menghitung jarak Euclidean antara setiap pasangan data. Fungsi pdist() menghitung jarak antar seluruh titik data, sedangkan squareform() mengubah hasil perhitungan tersebut menjadi bentuk matriks sehingga jarak setiap pasangan data dapat dilihat dengan lebih mudah. hal tersebut kemudian dimasukkan ke dalam Dataframe menggunakan pandas, dengan baris dan kolom diberi label a sampai f.

hasil yang diperoleh berupa matriks jarak seperti pada gambar. Nilai yang semakin kecil menunjukkan bahwa kedua data semakin dekat atau semakin mirip berdasarkan posisi koordinatnya. Sebaliknya, nilai jarak yang lebih besar menunjukkan bahwa kedua data memiliki posisi yang semakin jauh. linkage() digunakan untuk melakukan proses Hierarchical Clustering. Parameter method='single' menunjukkan bahwa metode yang digunakan adalah Single Linkage. Pada metode ini, jarak antar dua kelompok ditentukan berdasarkan jarak terdekat antara anggota dari kedua kelompok. Setelah proses clustering selesai, fungsi dendrogram() digunakan untuk menampilkan proses penggabungan data dalam bentuk grafik. Sumbu X menunjukkan data yang digunakan, sedangkan sumbu Y menunjukkan jarak ketika proses penggabungan terjadi. Pada dendrogram Single Linkage, data d dan f bergabung terlebih dahulu karena memiliki jarak paling kecil, yaitu 0,5. Setelah itu data a dan b bergabung pada jarak sekitar 0,707. Selanjutnya proses penggabungan dilakukan berdasarkan pasangan atau kelompok dengan jarak terdekat. Hasil perhitungan linkage menunjukkan tahapan penggabungan pada jarak sekitar 0,5; 0,707; 1,0; 1,414; dan 2,5. Hal ini menunjukkan bahwa Single Linkage cenderung menghubungkan kelompok berdasarkan anggota yang memiliki jarak paling dekat.

```python
import pandas as pd
df = pd.read_csv("Mall_Customers.csv")
df.head()
```

```python
x = df.iloc[:, [3, 4]].values
```

```python
from sklearn.preprocessing import StandardScaler

data_scaler = StandardScaler()

scaled_data = data_scaler.fit_transform(x)
scaled_data. shape
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram

clustering = linkage(scaled_data, method="complete", metric="euclidean")
```

```python
import matplotlib.pyplot as plt

dendrogram(clustering)
plt.show()
```

```python
# TUGAS PRAKTIKUM
# SINGLE LINKAGE

# Single Linkage
linked_single = linkage(data, method='single', metric='euclidean')

plt.figure(figsize=(12, 8))
plt.title("Hierarchical Clustering Dendrogram - Single Linkage")
dendrogram(
    linked_single,
    labels=['a', 'b', 'c', 'd', 'e', 'f'],
    leaf_rotation=90
)

plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.show()
```

pada cell ini linkage() digunakan untuk melakukan proses Hierarchical Clustering. Parameter method='single' menunjukkan bahwa metode yang digunakan adalah Single Linkage. Pada metode ini, jarak antar dua kelompok ditentukan berdasarkan jarak terdekat antara anggota dari kedua kelompok. Setelah proses clustering selesai, fungsi dendrogram() digunakan untuk menampilkan proses penggabungan data dalam bentuk grafik. Sumbu X menunjukkan data yang digunakan, sedangkan sumbu Y menunjukkan jarak ketika proses penggabungan terjadi.
Pada dendrogram Single Linkage, data d dan f bergabung terlebih dahulu karena memiliki jarak paling kecil, yaitu 0,5. Setelah itu data a dan b bergabung pada jarak sekitar 0,707. Selanjutnya proses penggabungan dilakukan berdasarkan pasangan atau kelompok dengan jarak terdekat. Hasil perhitungan linkage menunjukkan tahapan penggabungan pada jarak sekitar 0,5; 0,707; 1,0; 1,414; dan 2,5. Hal ini menunjukkan bahwa Single Linkage cenderung menghubungkan kelompok berdasarkan anggota yang memiliki jarak paling dekat.

```python
# TUGAS PRAKTIKUM
# MEMBANDINGKAN SINGLE LINKAGE COMPLETE LINKAGE

# Complete Linkage
linked_complete = linkage(data, method='complete', metric='euclidean')

plt.figure(figsize=(12, 8))
plt.title("Hierarchical Clustering Dendrogram - Complete Linkage")
dendrogram(
    linked_complete,
    labels=['a', 'b', 'c', 'd', 'e', 'f'],
    leaf_rotation=90
)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.show()

```

pada cell ini kode ini hampir sama dengan Single Linkage, tetapi parameter method diubah menjadi 'complete'. Complete Linkage menentukan jarak antar kelompok berdasarkan jarak terjauh antara anggota dari dua kelompok. dengan demikian, metode ini mempertimbangkan seluruh anggota kelompok sehingga penggabungan cluster cenderung lebih memperhatikan jarak maksimum.
pada Complete Linkage, data d dan f tetap menjadi pasangan pertama yang bergabung pada jarak 0,5, kemudian a dan b bergabung pada jarak sekitar 0,707. Setelah itu, proses penggabungan berbeda dari Single Linkage karena Complete Linkage mempertimbangkan jarak terjauh antaranggota cluster. Nilai penggabungan berikutnya adalah sekitar 1,118, 2,5, dan akhirnya 5,657. Karena menggunakan jarak maksimum, dendrogram Complete Linkage memiliki ketinggian penggabungan akhir yang lebih besar dibandingkan Single Linkage.

```python
# TUGAS PRAKTIKUM
# AVERAGE LINKAGE

linked_average = linkage(data, method='average', metric='euclidean')

plt.figure(figsize=(12, 8))
plt.title("Hierarchical Clustering Dendrogram - Average Linkage")
dendrogram(
linked_average,
labels=['a', 'b', 'c', 'd', 'e', 'f'],
leaf_rotation=90

)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.show()
```

pada kode ini, metode clustering diubah menjadi 'average'. Average Linkage menentukan jarak antara dua kelompok berdasarkan rata-rata jarak antar seluruh pasangan data dari kedua kelompok. Dengan metode ini, penggabungan cluster tidak hanya memperhatikan jarak terdekat seperti Single Linkage atau jarak terjauh seperti Complete Linkage, tetapi menggunakan nilai rata-rata jarak. hasil Average Linkage menunjukkan bahwa data d dan f bergabung terlebih dahulu pada jarak 0,5, kemudian data a dan b bergabung pada jarak sekitar 0,707. Penggabungan berikutnya terjadi pada jarak sekitar 1,059, 2,050, dan 3,826. Hasil ini berada di antara karakteristik Single Linkage dan Complete Linkage karena menggunakan rata-rata jarak dalam menentukan penggabungan cluster.

```python
# Membandingkan ketiga dendogram
# Membuat clustering dengan tiga metode
linked_single = linkage(data, method='single', metric='euclidean')
linked_complete = linkage(data, method='complete', metric='euclidean')
linked_average = linkage(data, method='average', metric='euclidean')

# Menampilkan ketiga dendrogram
plt.figure(figsize=(18, 6))

plt.subplot(1, 3, 1)
dendrogram(linked_single, labels=['a', 'b', 'c', 'd', 'e', 'f'])
plt.title("Single Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")

plt.subplot(1, 3, 2)
dendrogram(linked_complete, labels=['a', 'b', 'c', 'd', 'e', 'f'])
plt.title("Complete Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")

plt.subplot(1, 3, 3)
dendrogram(linked_average, labels=['a', 'b', 'c', 'd', 'e', 'f'])
plt.title("Average Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")

plt.tight_layout()
plt.show()
```

kode tersebut menjalankan ketiga metode Hierarchical Clustering pada dataset yang sama, yaitu Single Linkage, Complete Linkage, dan Average Linkage. Hasil dari masing-masing metode kemudian ditampilkan dalam satu gambar agar struktur dendrogram dapat dibandingkan secara langsung. subplot(1, 3, ...) digunakan untuk membuat tiga grafik dalam satu tampilan, sehingga perbedaan proses penggabungan cluster dapat terlihat dengan lebih jelas. Ketiga dendrogram memiliki beberapa kesamaan pada tahap awal. Data d dan f bergabung terlebih dahulu karena memiliki jarak paling kecil, yaitu 0,5, sedangkan a dan b menjadi pasangan berikutnya dengan jarak sekitar 0,707. Perbedaan mulai terlihat pada tahap penggabungan cluster berikutnya. Single Linkage menghasilkan penggabungan pada jarak yang relatif lebih rendah karena menggunakan jarak terdekat. Complete Linkage menghasilkan jarak penggabungan yang lebih tinggi karena menggunakan jarak terjauh. Sementara itu, Average Linkage menghasilkan nilai di antara keduanya karena menggunakan rata-rata jarak.
