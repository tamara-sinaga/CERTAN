```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset Iris dari scikit-learn
iris = load_iris()
```

```python
#TUGAS PRAKTIKUM
#42324042_Tamara Sinaga

import numpy as np
import pandas as pd

from sklearn.datasets import load_iris
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Load dataset Iris
iris = load_iris()

# Membuat DataFrame
df = pd.DataFrame(iris.data, columns=iris. feature_names)
df["Species"] = iris.target_names[iris. target]

# Menentukan fitur don target
X = df.drop(columns=["Species"])
y = df["Species"]

# Encoding target
lab = LabelEncoder()
y = lab. fit_transform(y)

# Membagi data training dan testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=0
)

# Membuat dan melatih model Gaussian Naive Bayes
model = GaussianNB()
NB_model = model. fit(X_train, y_train)

# Prediksi data testing
result = NB_model.predict(X_test)

# Menghitung accuracy
accuracy = accuracy_score(y_test, result) * 100
print("Accuracy of data testing:", accuracy)

# Data baru sesuai tugas
input_data = pd.DataFrame(
    np.array([[7.3, 3.8, 4.6, 2.3]]),
    columns=iris. feature_names
    )

# Prediksi data baru
result = NB_model.predict(input_data)

# Mengubah angka prediksi menjadi nama species
prediction_label = lab.inverse_transform(result)

print("Hasil prediksi untuk input baru adalah:", prediction_label)
```

pada cell berikut algoritma Gaussian Naïve Bayes dapat digunakan untuk melakukan klasifikasi pada dataset Iris karena seluruh fitur yang digunakan merupakan data numerik kontinu. Model yang dilatih memperoleh akurasi pengujian sekitar 96,67%, sehingga dapat dikatakan memiliki kemampuan klasifikasi yang baik pada data Iris. Untuk tugas yang diberikan, yaitu melakukan prediksi terhadap data baru dengan nilai [7.3, 3.8, 4.6, 2.3], model menghasilkan prediksi virginica. Dengan demikian, jawaban dari tugas praktikum adalah bahwa data baru tersebut masuk ke dalam species Iris virginica
