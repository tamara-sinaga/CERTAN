```python
def f(x,y):
    return np.cos(np.exp(-x**2-y**2))*np.sin(np.sqrt(x**2+y**2))

x = np.linspace(-np.pi, np.pi, 30)
y = np. linspace(-np.pi, np.pi, 30)

X, Y = np.meshgrid(x,y)
Z = f(X,Y)

fig = plt.figure()
ax = plt.axes(projection = '3d' )
ax.contour3D(X, Y, Z, 50, cmap = 'binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
#Tugas_02_42324042

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2.5, 2.5, 800)
y = np.linspace(-2.5, 2.5, 800)
X, Y = np.meshgrid(x, y)

# Mendefinisikan persamaan f(x, y)
F = X ** 6 + 3*X ** 4*Y ** 2 -3*X ** 4 + 3*X ** 2*Y ** 4 - X ** 2*Y ** 3 - 6*X ** 2*Y ** 2 + 3*X ** 2 + Y ** 6 - 3*Y ** 4 + 3*Y ** 2

# Membuat plot contour untuk F(X, Y) = 1
plt.figure(figsize=(7,7))
plt.contour(X, Y, F, levels=[1], colors='red')

# Pengaturan tampilan grafik
plt.title(r'Grafik 2D Implisit: $x^6+3x^4y^2-3x^4+\dots=1$')
plt.xlabel('Sumbu x')
plt.ylabel('Sumbu Y')
plt.grid(True, linestyle='--', alpha=0.6)
plt.gca().set_aspect('equal', adjustable='box')

plt.tight_layout()
plt.show()
```

```python
#Tugas_03_42324042

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. membuat grid nilai x dan y
x = np.linspace(-2, 2,300)
y = np.linspace(-2, 2, 300)
X, Y= np.meshgrid(x, y)

# 2. menghitung nilai z sesuai persamaan
Z=(5*Y/4- np.abs(x))**2+X**2

# 3. menggambar permukaan 3D
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(x, Y, Z, cmap='plasma', edgecolor='none', antialiased=True)

ax.set_title(r'$z=\left(\frac{5y}{4}-|x|\right)^2+x^2$', fontsize=12)
ax. set_xlabel('x')
ax. set_ylabel('y')
ax. set_zlabel('z')
fig.colorbar(surf, shrink=0.6, aspect=12, label='z')

plt.tight_layout()
plt.show()
```
