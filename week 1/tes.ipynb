```python
#membuat list dengan for dan if
list1 =[i for i in range(0,20,2) if i !=2]
print("list 1 :", list1)

list2 =[i for i in range(0,20) if i%2 == 0]
print("list 2:", list2)

list3 = [i for i in range(0,20) if i%2 != 0]
print("list 3 :", list3)
```

```python
#cara advance membuat array
a = np.array( ([1,2,3], [4,5,6]), dtype = float ) 
print(a)

#membuat array dengan iterable
iterable = (x*2 for x in range(5))

a = np.fromiter(iterable, dtype = float)
print(a)

#multitype array
dtipe = [('nama', 'S255'), ('tinggi', int)]
data = [('Ricky Sitorus', 170),
        ('Risky Siahaan', 172),
        ('Joshua Silitonga', 180)]

a = np.array(data, type = dtipe)
print(a)
```
