```python
import pandas as pd

path = "GlobalWeatherRepository.csv"
df = pd.read_csv(path)

df .head()
```

```python
#NOMOR 1
#Tugas Analisis Korelasi temperature_farenheit

korelasi = df[
    [
        "temperature_fahrenheit",
        "pressure_mb",
        "air_quality_Sulphur_dioxide"
    ]
].corr()

korelasi
```

Penjelasan :
Temperature Fahrenheit vs Pressure memiliki nilai korelasi sebesar -0,486833, yang menunjukkan hubungan negatif sedang. Artinya, ketika suhu meningkat, tekanan udara cenderung menurun.
Temperature Fahrenheit vs Sulphur Dioxide memiliki nilai korelasi sebesar -0,023152, yang menunjukkan hubungan negatif sangat lemah. Nilai yang mendekati 0 menunjukkan bahwa hampir tidak terdapat hubungan linear antara suhu dan kadar Sulphur dioxide.
Pressure vs Sulphur Dioxide memiliki nilai korelasi sebesar 0,018620, yang menunjukkan hubungan positif sangat lemah. Artinya, perubahan tekanan udara hampir tidak memiliki hubungan linear dengan perubahan kadar Sulphur dioxide.

```python
#NOMOR 1
#Tugas Analisis Korelasi pressure_mb

korelasi_1 = df[
    ["temperature_fahrenheit", "pressure_mb"]
].corr()

korelasi_1
```

```python
#NOMOR 1
#Tugas Analisis Korelasi air_quality_Sulphur_dioxide

korelasi_3 = df[
    ["pressure_mb", "air_quality_Sulphur_dioxide"]
].corr()

korelasi_3
```

Penjelasan :
Nilai korelasi: 0,01862
Nilai korelasi mendekati 0, sehingga hubungan antara kedua variabel sangat lemah.
Nilai korelasi positif, sehingga arah hubungannya positif.
Hubungan positif tersebut menunjukkan bahwa ketika pressure_mb meningkat, air_quality_Sulphur_dioxide cenderung meningkat, tetapi pengaruhnya sangat kecil.
Dapat disimpulkan bahwa hampir tidak terdapat hubungan linear antara tekanan udara dan kadar Sulphur dioxide pada dataset tersebut.
Perubahan pressure_mb tidak banyak berkaitan dengan perubahan air_quality_Sulphur_dioxide.

```python
import pandas as pd
```

```python
df = pd.read_csv("shopping_trends.csv")

df .head()
```

```python
df.columns
```

```python
#NOMOR 2
#Tugas Analisa mean
mean = df["Previous Purchases"].mean()

print("Mean:", mean)
```

```python
#NOMOR 2
#ugas Analisa std
std = df["Previous Purchases"].std()

print("Std:", std)
```

```python
#Tugas Analisa Min
minimum = df["Previous Purchases"].min()

print("Min:", minimum)
```

```python
#Nomor 2 Tugas Analisa Q1
q1 = df["Previous Purchases"].quantile(0.25)

print("Q1:", q1)
```

```python
#Nomor 2 Tugas Analisa Max
maximum = df["Previous Purchases"].max()

print("Max:", maximum)
```

```python
#Nomor 2 Tugas Analisa Q2
q2 = df["Previous Purchases"].quantile(0.50)

print("Q2:", q2)
```

```python
#Nomor 2 Tugas Analisa Q3
q3 = df["Previous Purchases"].quantile(0.75)

print("Q3:", q3)
```

```python
#Analisa Korelasi Age vs Purchase Amount (USD)

korelasi_age_purchase = df[
["Age", "Purchase Amount (USD)"]
].corr()

korelasi_age_purchase
```

```python
#Analisa Korelasi Purchase Amount (USD) vs Review Rating

korelasi_purchase_review = df[
    ["Purchase Amount (USD)", "Review Rating"]
].corr()

korelasi_purchase_review
```

Penjelasan :
Data Previous Purchases memiliki jumlah 3.900 data. Nilai rata-rata (mean) sebesar 25,35 yang menunjukkan bahwa secara rata-rata setiap pelanggan telah melakukan sekitar 25 pembelian sebelumnya. Nilai standar deviasi (std) sebesar 14,45 menunjukkan bahwa jumlah pembelian sebelumnya cukup bervariasi dari nilai rata-ratanya. Nilai terendah (minimum) adalah 1 pembelian, sedangkan nilai tertinggi (maximum) adalah 50 pembelian.
Berdasarkan nilai kuartil, Q1 (25%) = 13, artinya sekitar 25% pelanggan memiliki jumlah pembelian sebelumnya sebanyak 13 atau kurang. Q2 (median) = 25, yang berarti 50% data memiliki jumlah pembelian sebelumnya 25 atau kurang, sedangkan 50% lainnya memiliki lebih dari 25 pembelian. Sementara itu, Q3 (75%) = 38, sehingga sekitar 75% pelanggan memiliki jumlah pembelian sebelumnya sebanyak 38 atau kurang.

Secara keseluruhan, data Previous Purchases berada pada rentang 1–50 pembelian, dengan nilai tengah 25 pembelian. Selisih antara Q1 dan Q3 adalah 25 pembelian (38 − 13), sehingga 50% data berada pada rentang 13 hingga 38 pembelian.
